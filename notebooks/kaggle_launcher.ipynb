# Kaggle launcher
Enable a GPU and internet in Notebook settings. Pin `GIT_REF` to the commit or tag used by the experiment. The dataset mounted below `DATA_ROOT` must preserve the manifest path stored in the recipe.

```python
import os, subprocess

REPOSITORY = "https://github.com/SiennaDai/benchmark-experiments.git"
GIT_REF = "main"  # For formal runs, replace with an immutable commit SHA or tag.
RECIPE = "recipes/mini_bf16_adamw.json"
RUN_ID = "adamw_trial_01"
CHECKOUT = "/kaggle/working/benchmark-experiments"

if not os.path.isdir(os.path.join(CHECKOUT, ".git")):
    subprocess.run(["git", "clone", REPOSITORY, CHECKOUT], check=True)
subprocess.run(["git", "-C", CHECKOUT, "fetch", "--tags", "origin"], check=True)
subprocess.run(["git", "-C", CHECKOUT, "checkout", "--detach", GIT_REF], check=True)
```

```python
env = os.environ.copy()
env.update({
    "DATA_ROOT": "/kaggle/input",
    "OUTPUT_ROOT": "/kaggle/working/runs",
    "MAX_WALL_SECONDS": "39600",  # Leave time to save Notebook output.
    "KAGGLE_INSTALL_DEPS": "0",  # Set to 1 only if preflight reports a missing add-on.
    # "RESUME_CHECKPOINT": "/kaggle/input/my-resume/run/checkpoints/latest.pt",
})
subprocess.run(["bash", "scripts/run_kaggle.sh", RECIPE, RUN_ID], cwd=CHECKOUT, env=env, check=True)
```
